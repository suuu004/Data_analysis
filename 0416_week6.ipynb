**과제 1**

숙대입구 대신 다른 역으로 바꾸어 결과 확인하기
-> 서울역으로 변경함

```python
import csv
import numpy as np

filename = 'subway_congestion.csv'

# 분석할 조건 설정
target_station = "서울역" # 역 변경
target_day = "평일"
target_direction = "상선"

# 결과 저장할 변수
times = None
congestion = None

# CSV 파일 열기
with open(filename, 'r', encoding='cp949') as f:
    reader = csv.DictReader(f)

    # 한 줄씩 확인하기
    for row in reader:
      if(
          row["출발역"] == target_station
          and row["요일구분"] == target_day
          and row["상하구분"] == target_direction
      ):
        # 앞의 5개 열은 설명 정보, 뒤는 시간대 열
        time_columns = list(row.keys())[5:]

        # 시간대 이름을 numpy 배열로 만들기
        times = np.array(time_columns)

        # 혼잡도 값을 숫자로 바꿔 numpy 배열로 만들기
        congestion_values = []

        for col in time_columns:
          if row[col] != '':
            congestion_values.append(float(row[col]))
          else:
            congestion_values.append(0.0)

        congestion = np.array(congestion_values)

        break


# 기본 통계 계산
print("평균 혼잡도:", np.mean(congestion))
print("최대 혼잡도:", np.max(congestion))
print("최소 혼잡도:", np.min(congestion))

# 가장 혼잡한 시간대 찾기
max_index = np.argmax(congestion)
print("가장 혼잡한 시간대:", times[max_index])

# 혼잡도 100 이상 시간대만 선택하기
print("혼잡도가 100 이상인 시간대:", times[congestion >= 100])

```

[실행 결과]

서울역의 여러 혼잡도 수치를 확인할 수 있다.

```python
import csv
import numpy as np

filename = 'subway_congestion.csv'

# 분석할 조건 설정
target_station = "서울역" # 역 변경
target_day = "평일"
target_direction = "상선"

# 결과 저장할 변수
times = None
congestion = None

# CSV 파일 열기
with open(filename, 'r', encoding='cp949') as f:
    reader = csv.DictReader(f)

    # 한 줄씩 확인하기
    for row in reader:
      if(
          row["출발역"] == target_station
          and row["요일구분"] == target_day
          and row["상하구분"] == target_direction
      ):
        # 앞의 5개 열은 설명 정보, 뒤는 시간대 열
        time_columns = list(row.keys())[5:]

        # 시간대 이름을 numpy 배열로 만들기
        times = np.array(time_columns)

        # 혼잡도 값을 숫자로 바꿔 numpy 배열로 만들기
        congestion_values = []

        for col in time_columns:
          if row[col] != '':
            congestion_values.append(float(row[col]))
          else:
            congestion_values.append(0.0)

        congestion = np.array(congestion_values)

        break


# 기본 통계 계산
print("평균 혼잡도:", np.mean(congestion))
print("최대 혼잡도:", np.max(congestion))
print("최소 혼잡도:", np.min(congestion))

# 가장 혼잡한 시간대 찾기
max_index = np.argmax(congestion)
print("가장 혼잡한 시간대:", times[max_index])

# [추가] 가장 한산한 시간대 찾기
min_index = np.argmin(congestion)
print(f"가장 한산한 시간대: {times[min_index]}")

# 혼잡도 100 이상 시간대만 선택하기
print("혼잡도가 100 이상인 시간대:", times[congestion >= 100])

```

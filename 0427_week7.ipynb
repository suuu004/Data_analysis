**과제 1**

두 CSV 파일 모두 사용하기

```python
import pandas as pd
df1 = pd.read_csv("/content/subway_congestion01.csv", encoding="cp949")
df2 = pd.read_csv("/content/subway_congestion02.csv", encoding="cp949")
print(df1.shape)
print(df1.head())
print(df2.shape)
print(df2.head())
```

[실행 결과]

각 csv 파일들을 읽고, df1와 df2의 내용을 확인할 수 있다.

**과제 2**

관심 있는 역 2개를 선택, 평일 데이터 사용

-> 시청, 서울역 선택함

```python
# df1, df2 합치기
df = pd.concat([df1, df2])

# 시청, 서울역 선택

target1 = df[(df["출발역"] == "시청") & (df['요일구분']=="평일")]
small1 = target1[["호선", "출발역","상하구분","8시00분","18시00분"]]
print(small1.head())

target2 = df[(df["출발역"] == "서울역") & (df['요일구분']=="평일")]
small2 = target2[["호선", "출발역","상하구분","8시00분","18시00분"]]
print(small2.head())
```

[실행 결과]

처음에 불러온 두 csv 파일을 합치고, 지정한 출발역이 시청과 서울역인 정보들이 각각 출력되는 것을 확인할 수 있다.

**과제 3**

관심 2개 시간대 평균 혼잡도 비교(예: 8시 00분 & 18시 00분)

```python
weekday = df[df['요일구분'] == '평일']
station_mean = weekday.groupby('출발역')[['8시00분','18시00분']].mean() # 역별, 8시 18시 혼잡도 구분
display(station_mean)
```

[실행 결과]

groupby를 통해 각 역별로 시간별 혼잡도가 구분되어 출력되는 것을 확인할 수 있고, 그 평균 혼잡도를 알 수 있다.

**과제 4**

결과를 출력하고 CSV 파일로 저장하기

```python
top10 = station_mean.sort_values('8시00분', ascending=False).head(10)
display(top10) # 혼잡도 순으로 내림차순 정리

result = top10.reset_index()
result.to_csv('top_station_8am.csv', index=False, encoding= 'cp949') # csv로 파일 변환
display(result)
```

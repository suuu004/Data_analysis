**과제 2**

파일 열고 첫 5줄 출력하기

```python
import csv

f= open('/content/서울교통공사_지하철혼잡도정보_20251130.csv', "r", encoding="cp949")
r= csv.reader(f)

row_count =0
for row in r:
  print(row)
  row_count +=1
  if row_count == 5:  # 첫 5줄 확인
    break

f.close()
```

**과제 3**

파일에서 특정역에 해당하는 레코드만 출력하기
-> 출발역이 성수역, 시청역, 서울역인 데이터의 레코드 출력하게 수정했음

```python
import csv

filename = "/content/서울교통공사_지하철혼잡도정보_20251130.csv" # 업로드한 파일명
target_subway_station = {"시청", "서울역", "성수"}    # 칮으려는 역

count = 0

with open(filename, "r", encoding="cp949") as f:
  r = csv.DictReader(f)

  for row in r:
      if row['출발역'] in target_subway_station:
          count += 1
          print(row)

print(f"총 {count}개의 레코드를 찾았습니다.")
```

[실행 결과]

지정한 코드대로 출발역이 서울역, 시청, 성수인 레코드만 출력이 되는 것을 확인할 수 있다.

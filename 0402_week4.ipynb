**과제 1**

시간대별 승객 수 리스트를 본인이 직접 5개 값(임의값)으로 바꿔 입력 후, 혼잡도 결과를 2회 실행 비교

```python
def congestion_rate(passengers, capacity):
  return (passengers / capacity) * 100

times = ["06시", "07시", "08시", "09시", "10시"]
passengers = [160, 100, 310, 130, 230]  # 임의값 변경 1
capacity = 160

i=0
while i < len(times):
  rate = congestion_rate(passengers[i], capacity)
  print(f"{times[i]} 혼잡도: {rate:.1f}%")
  i = i + 1
```

**과제 2**

혼잡도 분류 기준 변경 후 실행 결과 비교하기

```python
passengers = [50, 100, 150, 200, 300] # 승객 수 초기화

def level(rate):
  if rate >= 150:  # 변경 전
    return "매우 혼잡"
  elif rate >= 100: # 변경 전
    return "혼잡"
  elif rate >= 70:  # 변경 전
    return "보통"
  else:
    return "여유"

max_rate = -1
max_time = ""

i = 0
while i < len(times):
  rate = congestion_rate(passengers[i], capacity)
  if rate > max_rate:
    max_rate = rate
    max_time = times[i]
  print(times[i], level(rate))
  i = i + 1

print("가장 혼잡한 시간:", max_time, f"({max_rate: .1f}%)")
```

```python
def level(rate):
  if rate >= 180:  # 매우 혼잡 기준 변경(150% -> 180%)
    return "매우 혼잡"
  elif rate >= 130: # 혼잡 기준 변경(100% -> 130%)
    return "혼잡"
  elif rate >= 95:  # 보통 기준 변경(70% -> 95%)
    return "보통"
  else:
    return "여유"

max_rate = -1
max_time = ""

i = 0
while i < len(times):
  rate = congestion_rate(passengers[i], capacity)
  if rate > max_rate:
    max_rate = rate
    max_time = times[i]
  print(times[i], level(rate))
  i = i + 1

print("가장 혼잡한 시간:", max_time, f"({max_rate: .1f}%)")
```

**실행 결과**

혼잡도 등급 분류 기준(매우 혼잡 기준 150% → 180% 등)을 상향 조정한 결과, 동일한 승객 수임에도 불구하고 일부 시간대의 등급이 '보통'에서 '여유' 등으로 하향 평가되는 등 전체적인 혼잡도 판정이 완화되었다. 하지만 등급 기준 변경은 수치 계산식에 영향을 주지 않으므로, 가장 혼잡한 시간대(10시)와 실제 혼잡도(187.5%) 등의 분석 데이터 자체는 변함없이 유지됨을 확인했다.

**과제 3**

정원을 180, 200으로 변경 후 최고 혼잡 시간과 등급 실행 결과 비교하기

```python
def congestion_rate(passengers, capacity):
    return (passengers / capacity) * 100

def level(rate):
    if rate >= 150:
        return "매우 혼잡"
    elif rate >= 100:
        return "혼잡"
    elif rate >= 70:
        return "보통"
    else:
        return "여유"

times = ["06시", "07시", "08시", "09시", "10시"]
passengers = [120, 210, 260, 230, 150]
capacity = 180  # 과제 3: 정원 180으로 변경

max_rate = -1
max_time = ""

print(f"--- [정원 {capacity}명 기준 분석 결과] ---")
i = 0
while i < len(times):
    rate = congestion_rate(passengers[i], capacity)
    if rate > max_rate:
        max_rate = rate
        max_time = times[i]
    print(f"{times[i]} 혼잡도: {rate:.1f}% | 상태: {level(rate)}")
    i = i + 1

print("-" * 30)
print("가장 혼잡한 시간:", max_time, f"({max_rate:.1f}%)")
```

**과제 1**

최근 3~4학기 GPA 리스트를 본인 성적으로 입력해 평균 GPA를 계산하고, 장학금 유형을 바꿔가며 최소 기준 충족 여부를 2회 실행 비교하기


<실행 1 : 성적우수 유형>



```python
# 장학금 유형별 최소 GPA (딕셔너리)
min_gpa_by_type = {
    "성적우수" : 3.70,
    "복지장학" : 2.50
}

# 고정 기준(튜플) - (최소학점, 최소봉사시간)
min_credits, min_volunteer = (12, 20)

# 결격 성적(집합)
bad_grades = {"F", "NP"}

# 장학 유형 입력
sch_type = input("장학금 유형 입력(성적우수/복지장학): ")

# GPA 4학기 입력 -> 리스트
g1 = float(input("1학기 GPA: "))
g2 = float(input("2학기 GPA: "))
g3 = float(input("3학기 GPA: "))
g4 = float(input("4학기 GPA: ")) # 한 학기 추가함
gpas = [g1, g2, g3, g4]

credits = int(input("이번 학기 이수학점: "))
volunteer = int(input("봉사시간: "))
discipline = input("징계 여부(y/n): ").strip() # y면 결격
has_bad = input("F 또는 NP가 있나요?(y/n): ").strip()

# 평균 GPA 계산(함수 없이)
total = 0.0
count = 0
for x in gpas:
  total = total + x
  count = count + 1
avg_gpa = total/ count

print(f"\n평균 GPA: {avg_gpa: .2f}")

# 장학금 유형 체크(딕셔너리 키 확인)
if sch_type not in min_gpa_by_type:
  print("지원하지 않는 장학금 유형입니다.")
else:
  min_gpa = min_gpa_by_type[sch_type]

  # 결격 여부 만들기
  is_disciplined = (discipline == "y")
  has_bad_grade = (has_bad == "y")

  # 조건을 if로 분기
  if sch_type == "성적우수":
    need_credits = 15
  else:
    need_credits = min_credits

  # 최종 판정(논리연산 이용)
  if(avg_gpa >= min_gpa) and (credits >= need_credits) and (not is_disciplined) and (not has_bad_grade):
    print("결과: 장학금 자격 있음")
  else:
    print("결과: 장학금 자격 없음")
    # 탈락 사유(단순 출력)
    if avg_gpa < min_gpa:
      print(f"- 평균 GPA 미달 (기준: {min_gpa: .2f})")
    if credits < need_credits:
      print(f"- 이수학점 미달 (기준: {need_credits}학점)")
    if is_disciplined:
      print("- 징계 이력 있음")
    if has_bad_grade:
      print("- N/FP 포함")
```

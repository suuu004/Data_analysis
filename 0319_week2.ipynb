**과제 1**

```python
# 공공 데이터 수치
rent = 650000
cpi = 108.2
food_base = 350000 # 식비 변경 1

# 개인 지출 입력
transport = int(input("월 교통비(원): "))
telecom = int(input("월 통신비(원): "))
leisure = int(input("월 여가비(원): "))
#food = int(input("월 식비(원): "))

# 계산
adj_food = food_base * (cpi/100)
total = rent + adj_food + transport + telecom + leisure

# 출력
print(f"보정 식비: {adj_food:,.0f}원")
print(f"예상 총생활비: {total:,.0f}원")

# 조건문 분류
if total < 1200000:
  print("생활비 수준: 낮음")
elif total < 1700000:
  print("생활비 수준: 보통")
else:
  print("생활비 수준: 높음")
```

**과제 2**

< 분류 기준 조정 >
1. 생활비 수준(낮음)
- 100000원으로 낮춤

2. 생활비 수준(보통)
- 1800000원으로 높임

```python
rent = 650000
cpi = 108.2
food_base = 400000

# 개인 지출 입력
transport = int(input("월 교통비(원): "))
telecom = int(input("월 통신비(원): "))
leisure = int(input("월 여가비(원): "))

# 계산
adj_food = food_base * (cpi/100)
total = rent + adj_food + transport + telecom + leisure

# 출력
print(f"보정 식비: {adj_food:,.0f}원")
print(f"예상 총생활비: {total:,.0f}원")

# 조건문 분류
if total < 1000000:  # 분류 기준 수정 1
  print("생활비 수준: 낮음")
elif total < 1800000:   # 분류 기준 수정 2
  print("생활비 수준: 보통")
else:
  print("생활비 수준: 높음")
```

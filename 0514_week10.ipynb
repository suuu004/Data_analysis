관심있는 공공데이터 csv 1개를 선택하고 프롬프팅을 활용하 데이터분석하기

```python
## 공공 데이터 구조 확인

# csv 파일 읽기
import pandas as pd

df = pd.read_csv("data.csv", encoding="cp949")  # 파일 이름 간단히 변경했음 (경찰청 서울특별시경찰청_관서별 성매매 사범 단속 현황_20251231.csv)
display(df.head())
```

```python
## 조건 필터링
# 2. 2016년 이후 데이터 필터링 및 구조 파악
years_cols = [f'{year}년' for year in range(2016, 2026)]
selected_cols = ['관서명'] + years_cols
df_filtered = df[selected_cols].copy()

# 3. 결측치 확인
print("결측치 확인:\n", df_filtered.isnull().sum())

## 평균 계산
# 4. 관서명별 2016~2025년 단속 평균 구하기
df_filtered['단속평균'] = df_filtered[years_cols].mean(axis=1)

# 5. 평균 기준 내림차순 정렬 및 상위 4개 추출
result_top4 = df_filtered.sort_values(by='단속평균', ascending=False).head(4)
print("\n상위 4개 관서:\n", result_top4[['관서명', '단속평균']])
```

[실행 결과]

서울청 본청이 광역 단위 단속을 주도하여 평균치가 가장 높으며, 경찰서 단위에서는 수서, 관악, 강서 순으로 성매매 단속 활동이 집중되어 있음을 확인할 수 있다.
